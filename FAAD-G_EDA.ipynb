```python
# Loading data
df = pd.read_csv('EDA_Data.csv', encoding='latin1')
```

```python
# Calculate the median of the Amount column
amount_median = df['Amount'].median()
print(f"The median of the 'Amount' column is: {amount_median}")

# Fill the missing values in the 'Amount' column with its median
df['Amount'] = df['Amount'].fillna(amount_median)

# Verify that there are no more missing values
print("Missing values remaining:")
print(df.isna().sum())
```
